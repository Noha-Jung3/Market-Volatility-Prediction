```python
from sqlalchemy import create_engine
import yfinance as yf
```

Connect python to PostgreSQL

```python
engine = create_engine("postgresql://postgres:VolatileMarket@localhost:5432/market_volatility")
```

```python
engine
```

Upload existing datasets 

```python
import pandas as pd

train = pd.read_csv("C:/Users/jungn/OneDrive/Documents/Projects/Market-Volatility-Prediction/data/train.csv")
```

```python
train.head()
```

```python
train.to_sql(
    "market_data",
    engine,
    if_exists="replace",
    index = False
)
```

```python
val = pd.read_csv("C:/Users/jungn/OneDrive/Documents/Projects/Market-Volatility-Prediction/data/validation.csv")
test = pd.read_csv("C:/Users/jungn/OneDrive/Documents/Projects/Market-Volatility-Prediction/data/test_set.csv")
```

```python
test = test.drop([0,1])
test = test.rename(columns={"Price":"Date"})
```

```python
test
```

create new column "dataset" to display where the data came from (test set, validation set or training set)

```python
train["dataset"] = "train"
val["dataset"] = "validation"
test["dataset"] = "test"
```

```python
train.head()
```

combine everything together

```python
all_data = pd.concat(
    [train,val,test],
    ignore_index=True
)
```

```python
all_data["dataset"].value_counts()
```

Create a table "market_data" containing the combined data

```python
all_data.to_sql(
    "market_data",
    engine,
    if_exists="replace",
    index = False
)
```

```python
all_data.columns.tolist()
```

cleanup

```python
all_data = all_data.drop(columns = ["Unnamed: 0"])

all_data["Date"] = pd.to_datetime(all_data["Date"])
```

```python
all_data
```

```python
all_data.to_sql(
    "market_data",
    engine,
    if_exists="replace",
    index=False
)
```

checking if our data is in the form we want:

```python
query = """
SELECT MAX("Date")
FROM market_data
"""

latest_date = pd.read_sql(
    query,
    engine
).iloc[0,0]

print(latest_date)
```

Download new market data from the latest date in the data up to current date using yfinance

```python
new_data = yf.download(
    "SPY",
    start=latest_date.strftime("%Y-%m-%d"),
    auto_adjust=True
)
```

```python
new_data.tail()
```

```python
new_data = new_data.reset_index()
```

```python
new_data
```

We make sure the whole historic dataset is not downloaded everytime, only the lastest date onwards.

```python
new_data = new_data[
    new_data["Date"] > latest_date
]
```

```python
print(new_data.head())
print(new_data.tail())
print(len(new_data))
```

checking the columns

```python
market_cols = pd.read_sql(
    """
    SELECT *
    FROM market_data
    LIMIT 1
    """,
    engine
)

print(market_cols.columns.tolist())
```

creating "raw_market_data" containing only OHLCV 

```python
raw_data = all_data[["Date","Open", "High", "Low", "Close", "Volume"]
].copy()
```

```python
raw_data.to_sql(
    "raw_market_data",
    engine,
    if_exists="replace",
    index = False
)
```
