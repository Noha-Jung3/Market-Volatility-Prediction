```python
query = """
SELECT MAX("Date")
FROM market_data
"""

latest_date = pd.read_sql(
    query,
    engine
).iloc[0,0]

print(latest_date)
```

```python
new_data = yf.download(
    "SPY",
    start=latest_date.strftime("%Y-%m-%d"),
    auto_adjust=True
)
```

```python
new_data.tail()
```

```python
new_data = new_data.reset_index()
```

```python
new_data
```

```python
new_data = new_data[
    new_data["Date"] > latest_date
]
```
